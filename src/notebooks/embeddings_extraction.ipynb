```python
import os
import sys
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from dotenv import find_dotenv, load_dotenv
from pathlib import Path
# Carica l'ambiente .env
load_dotenv(find_dotenv())

```

# Configurazione path directory src e import del path assoluto

```python
src_path = (Path.cwd() / "..").resolve()
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from app.semantic_clustering.embeddings import load_embeddings
```

```python
similarity_matrix = None
Data_Path = "../../" + os.environ.get("DATA_EMBEDDINGS_PATH") + "email_embeddings.npy"

embeddings = load_embeddings(Data_Path)

if 'embeddings' in locals() or 'embeddings' in globals():
    print("OK")
    similarity_matrix = cosine_similarity(embeddings)
else:
    print("❌ Errore: La variabile 'embeddings' non esiste.")
```

```python
          
print(f"Shape della matrice di similarità: {similarity_matrix.shape}")
print(f"Significato: È una mappa quadrata di {similarity_matrix.shape[0]}x{similarity_matrix.shape[1]} confronti.\n")

# Per evitare di intasare l'output con migliaia di righe, stampiamo un sottoinsieme
print("--- FOCUS: SIMILARITÀ TRA I PRIMI 5 DOCUMENTI ---")

sub_matrix = np.round(similarity_matrix[:5, :5], 4)

# DataFrame
df_vis = pd.DataFrame(
    sub_matrix, 
    columns=[f"Doc {i}" for i in range(5)], 
    index=[f"Doc {i}" for i in range(5)]
)
display(df_vis)

# Statistiche generali della distribuzione per capire se i testi sono vicini o lontani
print("\n--- STATISTICHE DELLA SIMILARITÀ COMPLESSIVA ---")
# Estraiamo solo i valori fuori dalla diagonale principale per non inquinare la media con gli 1.0
mask = ~np.eye(similarity_matrix.shape[0], dtype=bool)
all_scores = similarity_matrix[mask]

print(f"Similarità media nel dataset: {all_scores.mean():.4f}")
print(f"Similarità massima rilevata:  {all_scores.max():.4f}")
print(f"Similarità minima rilevata:   {all_scores.min():.4f}")
```
