## 2. Input processed e configurazioni

```python
processed = pd.read_parquet(BASE_CONFIG.processed_input_path)

config_rows = []
for config in MODEL_CONFIGS:
    registry = MODEL_REGISTRY[config.model_name]
    chunk_texts, planned_index = build_embedding_jobs(processed, config)
    config_rows.append(
        {
            "model_name": config.model_name,
            "expected_dimensions": registry["embedding_dimensions"],
            "max_sequence_length": registry["max_sequence_length"],
            "input_prefix": registry["input_prefix"],
            "row_count": len(processed),
            "planned_chunks": len(chunk_texts),
            "mean_chunks_per_email": planned_index["chunk_count"].mean(),
            "max_chunks_per_email": planned_index["chunk_count"].max(),
            "embeddings_output": str(config.embeddings_output_path.relative_to(PROJECT_ROOT)),
            "metadata_output": str(config.embedding_metadata_output_path.relative_to(PROJECT_ROOT)),
        }
    )

config_summary = pd.DataFrame(config_rows)
print(config_summary.to_string(index=False))
```

## 3. Generazione artefatti embeddings

La cella seguente genera solo gli artefatti mancanti. Se un file `.npy`, index e metadata esistono gia', il modello viene saltato per evitare lavoro e download non necessari.

```python
generation_rows = []
for config in MODEL_CONFIGS:
    artifacts_exist = (
        config.embeddings_output_path.exists()
        and config.embedding_index_output_path.exists()
        and config.embedding_metadata_output_path.exists()
    )
    if artifacts_exist:
        metadata = json.loads(config.embedding_metadata_output_path.read_text(encoding="utf-8"))
        metadata["generation_status"] = "reused_existing_artifacts"
        generation_rows.append(metadata)
        continue

    start = time.perf_counter()
    metadata = run_embedding_pipeline(env_file=ENV_PATH, config=config)
    metadata["generation_seconds"] = round(time.perf_counter() - start, 3)
    metadata["generation_status"] = "generated"
    generation_rows.append(metadata)

generation_summary = pd.DataFrame(generation_rows)
print(generation_summary[[
    "model_name",
    "generation_status",
    "row_count",
    "embedding_dimensions",
    "chunk_count",
    "batch_size",
]].to_string(index=False))
```

```python
analysis_rows = []
per_model_details = {}

for config in MODEL_CONFIGS:
    embeddings = np.load(config.embeddings_output_path)
    index = pd.read_parquet(config.embedding_index_output_path)
    metadata = json.loads(config.embedding_metadata_output_path.read_text(encoding="utf-8"))
    norms = np.linalg.norm(embeddings, axis=1) if embeddings.size else np.array([])
    chunk_counts = index["chunk_count"] if not index.empty else pd.Series(dtype="int64")

    analysis_rows.append(
        {
            "model_name": config.model_name,
            "row_count": int(embeddings.shape[0]),
            "embedding_dimensions": int(embeddings.shape[1]) if embeddings.size else 0,
            "chunk_count": int(metadata.get("chunk_count", 0)),
            "mean_norm": float(norms.mean()) if norms.size else 0.0,
            "std_norm": float(norms.std()) if norms.size else 0.0,
            "min_norm": float(norms.min()) if norms.size else 0.0,
            "max_norm": float(norms.max()) if norms.size else 0.0,
            "mean_chunks_per_email": float(chunk_counts.mean()) if len(chunk_counts) else 0.0,
            "max_chunks_per_email": int(chunk_counts.max()) if len(chunk_counts) else 0,
        }
    )
    per_model_details[config.model_name] = {
        "embeddings": embeddings,
        "index": index,
        "metadata": metadata,
        "norms": norms,
        "chunk_counts": chunk_counts,
    }

analysis_summary = pd.DataFrame(analysis_rows)
print(analysis_summary.to_string(index=False))
```

```python
single_fig, axes = plt.subplots(len(MODEL_CONFIGS), 2, figsize=(13, 4 * len(MODEL_CONFIGS)))
if len(MODEL_CONFIGS) == 1:
    axes = np.array([axes])

for row_idx, config in enumerate(MODEL_CONFIGS):
    details = per_model_details[config.model_name]

    norms = details["norms"]
    if norms.size and np.isclose(norms.min(), norms.max()):
        axes[row_idx, 0].bar([f"{norms.mean():.4f}"], [len(norms)])
    else:
        axes[row_idx, 0].hist(norms, bins=30)
    axes[row_idx, 0].set_title(f"{slug}: norme L2")
    axes[row_idx, 0].set_xlabel("norma")
    axes[row_idx, 0].set_ylabel("email")

    chunk_distribution = details["chunk_counts"].value_counts().sort_index()
    axes[row_idx, 1].bar(chunk_distribution.index.astype(str), chunk_distribution.values)
    axes[row_idx, 1].set_title(f"{slug}: chunk per email")
    axes[row_idx, 1].set_xlabel("chunk_count")
    axes[row_idx, 1].set_ylabel("email")

single_fig.tight_layout()
single_fig.savefig(FIGURES_PATH / "embedding_single_model_diagnostics.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 5. Confronto tra modelli

## 6. Metriche di qualita' clustering (KMeans baseline)
Per valutare la qualita' intrinseca degli embeddings, applichiamo KMeans con k=2..10 e calcoliamo metriche clustering:
- **Silhouette Score:** misura quanto ogni punto e' coeso nel suo cluster vs separato dagli altri (-1 a 1, piu' alto e' meglio).
- **Davies-Bouldin Index:** rapporto dispersione intra-cluster / separazione inter-cluster (piu' basso e' meglio).
- **Calinski-Harabasz Index:** rapporto between-cluster vs within-cluster variance (piu' alto e' meglio).
Nota: KMeans non e' l'algoritmo ideale per dati semantici ad alta dimensionalita', ma fornisce una baseline diagnostica veloce.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
clustering_results = {}
for config in MODEL_CONFIGS:
    embeddings = per_model_details[config.model_name]["embeddings"]
    X = embeddings
    scores = []
    for k in range(2, 11):
        try:
            km = KMeans(n_clusters=k, random_state=42, n_init=10, verbose=0)
            labels = km.fit_predict(X)
            sil = silhouette_score(X, labels)
            db = davies_bouldin_score(X, labels)
            ch = calinski_harabasz_score(X, labels)
            scores.append({
                'k': k,
                'silhouette': float(sil),
                'davies_bouldin': float(db),
                'calinski_harabasz': float(ch),
            })
        except Exception as e:
            print(f"Errore per {config.model_name} k={k}: {e}")
            scores.append({'k': k, 'error': str(e)})
    clustering_results[model_slug] = scores
# Stampa tabella riepilogativa per ogni modello
for model_slug, scores in clustering_results.items():
    scores_df = pd.DataFrame(scores)
    print(f"\n{model_slug}:")
    print(scores_df.to_string(index=False))
```

```python
# Crea dataframe consolidato per confronto
all_scores = []
for model_slug, scores in clustering_results.items():
    for score_dict in scores:
        score_dict_copy = score_dict.copy()
        score_dict_copy['model_slug'] = model_slug
        all_scores.append(score_dict_copy)
clustering_df = pd.DataFrame(all_scores)
# Visualizza metriche clustering per tutti i modelli
clustering_fig, axes = plt.subplots(1, 3, figsize=(17, 5))
# Silhouette
for model_slug in clustering_df['model_slug'].unique():
    subset = clustering_df[clustering_df['model_slug'] == model_slug]
    axes[0].plot(subset['k'], subset['silhouette'], marker='o', label=model_slug)
axes[0].set_title('Silhouette Score vs k')
axes[0].set_xlabel('k (numero cluster)')
axes[0].set_ylabel('Silhouette Score')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
# Davies-Bouldin
for model_slug in clustering_df['model_slug'].unique():
    subset = clustering_df[clustering_df['model_slug'] == model_slug]
    axes[1].plot(subset['k'], subset['davies_bouldin'], marker='o', label=model_slug)
axes[1].set_title('Davies-Bouldin Index vs k (minore è migliore)')
axes[1].set_xlabel('k (numero cluster)')
axes[1].set_ylabel('Davies-Bouldin Index')
axes[1].legend()
axes[1].grid(True, alpha=0.3)
# Calinski-Harabasz
for model_slug in clustering_df['model_slug'].unique():
    subset = clustering_df[clustering_df['model_slug'] == model_slug]
    axes[2].plot(subset['k'], subset['calinski_harabasz'], marker='o', label=model_slug)
axes[2].set_title('Calinski-Harabasz Index vs k (maggiore è migliore)')
axes[2].set_xlabel('k (numero cluster)')
axes[2].set_ylabel('Calinski-Harabasz Index')
axes[2].legend()
axes[2].grid(True, alpha=0.3)
clustering_fig.tight_layout()
clustering_fig.savefig(FIGURES_PATH / "embedding_clustering_metrics.png", dpi=150, bbox_inches="tight")
plt.show()
# Salva i risultati clustering in JSON per reference
clustering_json_path = FIGURES_PATH / "embedding_clustering_metrics.json"
clustering_json_path.write_text(json.dumps(clustering_results, indent=2), encoding="utf-8")
print(f"Metriche clustering salvate in: {clustering_json_path}")
```

## 7. Analisi sommaria della qualita' embedding (riassunto)
Sulla base dell'analisi sopra:
- **Norme vettoriali:** Tutti i modelli producono vettori L2-normalizzati (norma ~1.0), come atteso dopo la pipeline di aggregazione e normalizzazione.
- **Similarita' pairwise:**
  - BGE ha similarita' media ~0.666 (vettori moderatamente vicini).
  - MiniLM ha similarita' media ~0.207 (vettori dispersi).
  - E5 ha similarita' media ~0.792 (vettori molto vicini).
  Interpretazione: MiniLM e' piu' sparse (discriminativo), E5 e' piu' dense (collassato). Nessuno e' ideale a priori senza valutazione downstream.
- **Metriche clustering (KMeans):**
  - Tutti i modelli mostrano silhouette score <0.05 per k=2..10, indicando scarsa separazione con KMeans.
  - Davies-Bouldin index ~3-4 e Calinski-Harabasz index ~8-14 sono tipicamente bassi.
  - Questo suggerisce che il dataset non ha classi semantiche forti o che KMeans non e' adatto a questa rappresentazione.
**Raccomandazione:** Procedere con algoritmi density-based (HDBSCAN) e/o riduzione dimensionale (UMAP) prima di clustering finale. KMeans baseline mostra che serve ulteriore analisi e preprocessing.

## 8. Conclusioni operative
### Sintesi risultati
Questo notebook produce una valutazione multi-dimensionale dei tre modelli embeddings:
1. **Proprietà intrinseche:**
   - Dimensionalità: MiniLM/BGE (384), E5 (768).
   - Copertura: Identica (355 email, ~545 chunk).
   - Normalizzazione: Tutti a L2 (norma ~1.0).
   - Dispersione vettoriale: MiniLM > BGE >> E5.
2. **Qualità clustering (baseline KMeans):**
   - Silhouette score: Bassissimo per tutti (<0.05), suggerendo cluster non separati con KMeans.
   - Davies-Bouldin: ~3-4 (valori più bassi sarebbero migliori).
   - Calinski-Harabasz: ~8-14 (valori più alti sarebbero migliori).
   - Conclusione: KMeans non e' ideale; serve HDBSCAN + UMAP o altro approccio.

- Implementare UMAP (riduzione a 50 o 2D) + HDBSCAN per clustering più robusto su ciascun modello.
- Se retrieval e' rilevante (e.g., per RAG), costruire gold set di query + relevant documento e valutare Recall@k/MRR.
- Scegliere modello sulla base di task downstream, non solo su metriche intrinseche.
- Se è necessario miglioramento di dominio, considerare fine-tuning contrastivo con dati in-domain.
